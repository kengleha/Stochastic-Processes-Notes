```python
import numpy as np
import matplotlib.pyplot as plt

# Define standard deviation
sd = 2

# Generate x-axis values
xa = np.arange(-6, 6.1, 0.1)

# Define normal distribution function
def normal_pdf(x, mu=0, sigma=sd):
    return 1 / (np.sqrt(2 * np.pi) * sigma) * np.exp(-(x - mu)**2 / (2 * sigma**2))

# Calculate f_x(x)
fx = normal_pdf(xa)

# Filter y-axis values for positive values
ya = xa[xa > 0]

# Define function to calculate f_y(y) for different n values
def f_y(y, n):
    return np.real(1 / (np.sqrt(2 * np.pi) * sd * np.abs(n * y**(n-1)/n)) * np.exp(-(y**(2/n)) / (2 * sd**2)))

# Calculate f_y(y) for odd and even n values
fyo = np.zeros_like(ya)
fye = np.zeros_like(ya)
for n in range(1, 5):
    if n % 2 == 0:
        fye[n//2] = f_y(ya, n)
    else:
        fyo[(n+1)//2] = f_y(ya, n)

# Plot all functions
plt.figure()
plt.plot(xa, fx, label='f_x(x)')
plt.plot(ya, fyo, label='f_y(y), n=odd')
plt.plot(ya, fye, label='f_y(y), n=even')
plt.xlabel('X / Y')
plt.ylabel('f_x(x) / f_y(y)')
plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Define the function
def f(tau):
    if tau == 0:
        return 1
    else:
        return np.sin(np.pi * tau) / (np.pi * tau)

# Generate values for tau
tau_values = np.linspace(0.001, 2, 1000)  # Avoid division by zero at tau = 0

# Compute the function values
function_values = [f(t) for t in tau_values]

# Plot the function
plt.figure(figsize=(8, 6))
plt.plot(tau_values, function_values, label=r'$\frac{\sin (\pi \tau)}{\pi \tau}$', color='b')
plt.title('Plot of $f(\\tau) = \\frac{\\sin (\\pi \\tau)}{\\pi \\tau}$')
plt.xlabel(r'$\tau$')
plt.ylabel(r'$f(\tau)$')
plt.grid(True)
plt.legend()
plt.show()
```
